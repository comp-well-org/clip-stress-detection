```python
import os
import numpy as np
import pandas as pd
from constant import EXPS_PATH, PMDATA_PATH, LIFESNAPS_PATH
```

```python
def print_pm_stats(unlabel_rate: int = 1):
    assert unlabel_rate in [1, 10, 50, 100]
    session = 'review/clip/quantile_global/seq:transformer_tab:resnet/'
    exclude = 'label'
    label_rate = 30
    exp_home_dir = os.path.join(
        EXPS_PATH,
        'pmdata',
        session,
        f'unlabel_rate_{unlabel_rate}',
        f'label_rate_{label_rate}',
        f'{exclude}',
    )
    aucs = []
    for fold in range(5):
        for seed in range(4):
            exp_path = os.path.join(exp_home_dir, f'fold_{fold}', f'seed_{seed}')
            zero_shot_results = pd.read_csv(os.path.join(exp_path, 'results.csv'))
            zero_shot_max = zero_shot_results['val_auc'].max()
            linear_prob_results = pd.read_csv(os.path.join(exp_path, 'results_lp.csv'))
            linear_prob_max = linear_prob_results['val_auc'].max()
            fine_tune_results = pd.read_csv(os.path.join(exp_path, 'results_ft.csv'))
            fine_tune_max = fine_tune_results['val_auc'].max()
            auc_max = max(zero_shot_max, linear_prob_max, fine_tune_max)
            aucs.append(auc_max)
            
    mu, sigma = np.mean(aucs), np.std(aucs)
    print(f'${mu*100:.2f} \pm {sigma*100:.2f}$')
    return aucs

_ = print_pm_stats(1)
_ = print_pm_stats(10)
_ = print_pm_stats(50)
_ = print_pm_stats(100)
```
