```python
import os
import numpy as np
import pandas as pd
from constant import EXPS_PATH
```

```python
def read_scores(dataset, flag='gen'):
    exp_path = os.path.join(EXPS_PATH, dataset, 'user', 'clip', flag)
    scores = []
    for fold in range(5):
        for seed in range(4):
            exp_path_fold = os.path.join(exp_path, f'fold_{fold}', f'seed_{seed}')
            # print(os.listdir(exp_path_fold))
            score = pd.read_csv(os.path.join(exp_path_fold, 'results_ft.csv'))
            score = score['val_acc'].max()
            scores.append(score)

    mu, sigma = np.mean(scores), np.std(scores)
    print(f'{dataset} {flag}')
    print(f'${mu*100:.2f} \pm {sigma*100:.2f}$')
    return scores

_ = read_scores('lifesnaps', 'gen')
_ = read_scores('lifesnaps', 'per')
_ = read_scores('pmdata', 'gen')
_ = read_scores('pmdata', 'per')
```
